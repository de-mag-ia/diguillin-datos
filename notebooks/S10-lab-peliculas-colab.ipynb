## K2 · Las dos reglas de k-means, escritas a mano

**Qué se pide.** Escriba las dos reglas del lunes como funciones de Python. **Asignar:** cada película se va con la estrella más cercana. **Actualizar:** cada estrella se muda al promedio de su grupo. Pruébelas con una sola ronda, partiendo con las 3 estrellas del lunes: ★1 en (2; 10), ★2 en (6; 10) y ★3 en (10; 6).

### 🔎 Cómo se resuelve

**La idea.** Las dos reglas del lunes, tal cual. `asignar()` calcula, para cada estrella, la **d²** de todas las películas a ella: (acción − x)² + (humor − y)². Cada película se queda con la estrella de menor d². `actualizar()` saca el promedio de acción y de humor de cada grupo: ahí se muda la estrella. `dibujar()` solo pinta.

**Línea por línea:**

| Código | Qué hace |
|---|---|
| `d2[k] = (peliculas.accion - ex) ** 2 + (peliculas.humor - ey) ** 2` | Una columna por estrella: la d² de **cada** película a esa estrella. Es la fórmula de la guía, para las 240 de una vez. |
| `grupo = d2.idxmin(axis=1)` | Por cada fila, **en qué columna** está la d² más chica: el número de la estrella más cercana (0, 1 o 2). |
| `inercia = d2.min(axis=1).sum()` | La d² de cada película a **su** estrella, sumada: la inercia. Mientras más chica, más apretados los grupos. |
| `peliculas.groupby(grupo)[["accion", "humor"]].mean()` | El promedio de acción y de humor de cada grupo: las estrellas nuevas. |
| `ESTRELLAS_INICIO = [(2, 10), (6, 10), (10, 6)]` | Las 3 estrellas del lunes. Python cuenta desde 0: la estrella 0 es ★1. |

````python
# K2 · Las dos reglas de k-means, escritas a mano

COLORES = ["#b8860b", "#2e7d32", "#c62828", "#1565c0"]

def dibujar(ax, grupo, estrellas, titulo):
    """Cada película del color de su estrella; cada estrella, una ★ grande de su color."""
    for k in range(len(estrellas)):
        sel = peliculas[grupo == k]
        ax.scatter(sel.accion, sel.humor, s=12, color=COLORES[k], alpha=0.7)
    for k, (ex, ey) in enumerate(estrellas):
        ax.scatter(ex, ey, marker="*", s=400, color=COLORES[k], edgecolor="black")
    ax.set_xlim(0, 10.5); ax.set_ylim(0, 10.5)
    ax.set_xlabel("Acción"); ax.set_ylabel("Humor"); ax.set_title(titulo)

def asignar(peliculas, estrellas):
    """Regla 1: cada película se va con la estrella más cercana."""
    d2 = pd.DataFrame()
    for k, (ex, ey) in enumerate(estrellas):
        d2[k] = (peliculas.accion - ex) ** 2 + (peliculas.humor - ey) ** 2
    grupo = d2.idxmin(axis=1)            # el número de la estrella más cercana
    inercia = d2.min(axis=1).sum()       # suma de las d² de cada película a SU estrella
    return grupo, inercia

def actualizar(peliculas, grupo):
    """Regla 2: cada estrella se muda al promedio de su grupo."""
    centros = peliculas.groupby(grupo)[["accion", "humor"]].mean()
    return centros.values.tolist()

# Una sola ronda, con las 3 estrellas del lunes
ESTRELLAS_INICIO = [(2, 10), (6, 10), (10, 6)]
grupo1, inercia1 = asignar(peliculas, ESTRELLAS_INICIO)
print("Películas por estrella:", grupo1.value_counts().sort_index().tolist())
print("Inercia de la ronda 1:", round(inercia1, 1))

estrellas2 = actualizar(peliculas, grupo1)
print("Las estrellas se mudan a:", [(round(x, 1), round(y, 1)) for x, y in estrellas2])

fig, ax = plt.subplots(figsize=(6, 6))
dibujar(ax, grupo1, ESTRELLAS_INICIO, "Ronda 1 · cada película elige su estrella")
plt.show()
````

## K3 · Repetir hasta que nadie cambie

**Qué se pide.** Ahora la vuelta completa: asignar → actualizar → asignar… hasta que **ninguna película cambie de estrella**. Anote en una tabla, por ronda, cuántas películas cambiaron y la inercia, y dibuje todas las rondas.

### 🔎 Cómo se resuelve

**La idea.** Un ciclo `for` que repite **asignar → actualizar** y en cada vuelta compara los grupos nuevos con los de la vuelta anterior. Cuando **ninguna película cambió**, corta con `break`. La tabla `historial` es la misma que el lunes llenaron a mano.

**Línea por línea:**

| Código | Qué hace |
|---|---|
| `for ronda in range(1, 21):` | Hasta 20 rondas. Por seguridad: siempre termina antes. |
| `cambios = int((grupo != grupo_anterior).sum())` | Compara película por película el grupo de ahora con el de la ronda anterior y cuenta cuántas cambiaron. |
| `if cambios == 0: break` | Nadie cambió: terminó (convergió). |
| `tabla.cambiaron.astype("Int64")` | Deja los cambios como números enteros. En la ronda 1 aparece `<NA>`: ahí no hay ronda anterior con qué comparar. |
| `estrellas = actualizar(peliculas, grupo)` | Si alguien cambió, las estrellas se mudan y se da otra vuelta. |
| `fig, ejes = plt.subplots(filas, 3, …)` | Un dibujo por ronda, 3 por fila, para ver cómo se mueven las estrellas. Los cuadros que sobran quedan en blanco. |

````python
# K3 · Repetir hasta que nadie cambie

estrellas = ESTRELLAS_INICIO
grupo_anterior = None
historial = []        # una fila por ronda
dibujos = []          # (grupo, estrellas) de cada ronda, para el gráfico

for ronda in range(1, 21):                        # 20 rondas como máximo, por seguridad
    grupo, inercia = asignar(peliculas, estrellas)
    if grupo_anterior is None:
        cambios = None                            # en la ronda 1 todas eligen por primera vez
    else:
        cambios = int((grupo != grupo_anterior).sum())
    historial.append({"ronda": ronda, "cambiaron": cambios, "inercia": round(inercia, 1),
                      "estrellas": [(round(x, 1), round(y, 1)) for x, y in estrellas]})
    dibujos.append((grupo, estrellas))
    if cambios == 0:                              # nadie cambió: terminó
        break
    estrellas = actualizar(peliculas, grupo)
    grupo_anterior = grupo

tabla = pd.DataFrame(historial).set_index("ronda")
tabla["cambiaron"] = tabla.cambiaron.astype("Int64")   # entero; en la ronda 1 queda <NA>
print(tabla)

rondas = len(tabla)
cambios_ronda2 = int(tabla.loc[2, "cambiaron"])
inercia_k3 = tabla.inercia.iloc[-1]
print("Películas por estrella al final:", grupo.value_counts().sort_index().tolist())

filas = (rondas + 2) // 3                         # 3 dibujos por fila
fig, ejes = plt.subplots(filas, 3, figsize=(13, 4.3 * filas))
ejes = ejes.flatten()
for i, (g, est) in enumerate(dibujos):
    dibujar(ejes[i], g, est, f"Ronda {i + 1}")
for ax in ejes[rondas:]:                          # los cuadros que sobran, en blanco
    ax.axis("off")
plt.tight_layout()
plt.show()
````

## K4 · Lo mismo con scikit-learn

**Qué se pide.** En el trabajo real no se escribe k-means a mano: se usa la librería **scikit-learn**. Córrala con **las mismas 3 estrellas iniciales** y compare película por película con su resultado.

### 🔎 Cómo se resuelve

**La idea.** `KMeans` hace las mismas dos reglas. Con `init=` le damos **nuestras** estrellas iniciales y con `n_init=1` le pedimos una sola corrida, para comparar justo.

**Línea por línea:**

| Código | Qué hace |
|---|---|
| `KMeans(n_clusters=3, init=…, n_init=1)` | k = 3, partiendo de las 3 estrellas del lunes, una sola vez. |
| `km.fit(X)` | Corre las rondas hasta que nadie cambie. |
| `(km.labels_ == grupo.values).sum()` | Cuántas películas quedaron en el mismo grupo con la librería y con nuestras funciones. |
| `km.inertia_` | La inercia final que calcula la librería. |

````python
# K4 · Lo mismo con scikit-learn

import numpy as np
from sklearn.cluster import KMeans

X = peliculas[["accion", "humor"]]
km = KMeans(n_clusters=3, init=np.array(ESTRELLAS_INICIO, dtype=float), n_init=1)
km.fit(X)

coinciden = int((km.labels_ == grupo.values).sum())
print("Películas en el mismo grupo que con nuestras funciones:", coinciden, "de", len(peliculas))
print("Estrellas finales de scikit-learn:", km.cluster_centers_.round(1).tolist())
print("Estrellas finales nuestras:       ", [[round(x, 1), round(y, 1)] for x, y in estrellas])
print("Inercia de scikit-learn:", round(km.inertia_, 1), " · nuestra:", inercia_k3)
````

## K5 · ¿Cuántos grupos? El codo

**Qué se pide.** El lunes usamos k = 3 porque lo decidimos nosotros. Pruebe k de 1 a 8 y grafique la inercia de cada uno: el **codo** (donde la curva deja de bajar fuerte) sugiere cuántos grupos hay.

### 🔎 Cómo se resuelve

**La idea.** Se corre k-means para cada k de 1 a 8 y se anota la inercia. Con más grupos la inercia siempre baja; lo que importa es **dónde deja de bajar fuerte**. Aquí `n_init=10` repite cada corrida con 10 inicios distintos y se queda con el mejor.

**Línea por línea:**

| Código | Qué hace |
|---|---|
| `KMeans(n_clusters=k, n_init=10, random_state=2026)` | k grupos; 10 inicios al azar; `random_state` fija el azar para que a todos les dé lo mismo. |
| `inercias[k] = round(modelo.inertia_)` | Guarda la inercia de cada k. |
| `plt.plot(…, marker="o")` | La curva del codo. |

````python
# K5 · ¿Cuántos grupos? El codo

inercias = {}
for k in range(1, 9):
    modelo = KMeans(n_clusters=k, n_init=10, random_state=2026).fit(X)
    inercias[k] = round(modelo.inertia_)
print(inercias)

plt.figure(figsize=(6, 4))
plt.plot(list(inercias.keys()), list(inercias.values()), marker="o")
plt.xlabel("k (cantidad de grupos)"); plt.ylabel("Inercia")
plt.title("El codo")
plt.grid(alpha=0.3)
plt.show()
````

## K6 · Los 4 grupos

**Qué se pide.** Corra k-means con el k que sugiere el codo (k = 4), calcule el **perfil** de cada grupo (cuántas películas y su acción y humor promedio) y dibújelos.

### 🔎 Cómo se resuelve

**La idea.** k-means con el k del codo. El **perfil** (promedio de acción y de humor de cada grupo) es lo que permite ponerle nombre a cada grupo.

**Línea por línea:**

| Código | Qué hace |
|---|---|
| `peliculas["grupo"] = km4.labels_` | Anota en la tabla el grupo de cada película. |
| `peliculas.groupby("grupo").agg(…)` | Por grupo: cuántas películas y su acción y humor promedio. |
| `dibujar(ax, peliculas.grupo, km4.cluster_centers_, …)` | Los 4 grupos y sus 4 estrellas. |

````python
# K6 · Los 4 grupos

km4 = KMeans(n_clusters=4, n_init=10, random_state=2026).fit(X)
peliculas["grupo"] = km4.labels_

perfil = peliculas.groupby("grupo").agg(
    peliculas=("accion", "size"),
    accion=("accion", "mean"),
    humor=("humor", "mean"),
).round(1)
print(perfil)

fig, ax = plt.subplots(figsize=(6, 6))
dibujar(ax, peliculas.grupo, km4.cluster_centers_, "k = 4")
plt.show()
````

````python
# ✔ Verificación · K6
verificar("tamanos_k4", sorted(perfil.peliculas.tolist()))
````
