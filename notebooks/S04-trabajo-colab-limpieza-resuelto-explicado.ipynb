````python
# ============================================================
#  CELDA 0 · ejecutar primero, siempre
# ============================================================
import os
import pandas as pd
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", 30)
pd.set_option("display.width", 160)

# Los CSV crudos están publicados: se leen desde la URL, no hay que subir nada.
RUTA_DATOS = "https://raw.githubusercontent.com/de-mag-ia/diguillin-datos/main/crudo/"
clientes = pd.read_csv(RUTA_DATOS + "clientes.csv")
boletas  = pd.read_csv(RUTA_DATOS + "boletas.csv")

# Los números que deben salir en cada paso (los mismos que vio en Excel el 01/09).
ESPERADO = {
    'filas_clientes': 1245,
    'filas_boletas': 57527,
    'comunas_distintas': 21,
    'medios_pago_distintos': 4,
    'comunas_vacias': 87,
    'canal_alta_vacios': 49,
    'faltantes_despues': 0,
    'comuna_sin_dato': 87,
    'rut_repetidos_exactos': 30,
    'rut_repetidos_normalizados': 45,
    'clientes_finales': 1200,
    'boletas_repetidas_por_id': 60,
    'boletas_finales': 57467,
    'devoluciones': 70,
    'boletas_ventas': 57397,
    'boletas_sin_tarjeta': 22812,
    'huerfanas': 90,
    'dia_mas_boletas': 'sábado',
    'boletas_sabado': 12658,
    'edad_mediana': 40,
    'archivo_guardado': True,
    'huerfanas_keep_first': 90,
    'huerfanas_keep_last': 1730,
    'boletas_de_fichas_eliminadas': 1640,
    'fechas_imposibles': 18,
    'modelo_filas': 1200,
    'modelo_fugados': 291,
    'zeror_aciertos': 909,
    'oner_tramo_aciertos': 909,
    'oner_mejor_aciertos': 909,
    'oner_atributos_que_ganan': 0,
    'clientes_60mas': 220,
    'fugados_60mas': 8,
    'clientes_18_29': 272,
    'fugados_18_29': 106,
}

_resultados = {}

def verificar(clave, valor):
    """Compara su resultado con el esperado e imprime ✅ o ❌."""
    esperado = ESPERADO[clave]
    if hasattr(valor, "item"):          # numpy → Python
        valor = valor.item()
    ok = (valor == esperado)
    _resultados[clave] = ok
    if ok:
        print(f"✅ {clave} = {valor}")
    else:
        print(f"❌ {clave} = {valor}   (debía salir {esperado})")
    return ok

def resumen():
    """Cuántas verificaciones están en ✅."""
    total, ok = len(ESPERADO), sum(_resultados.values())
    pendientes = [k for k in ESPERADO if not _resultados.get(k)]
    print(f"{ok} de {total} verificaciones en ✅")
    if pendientes:
        print("Pendientes o con ❌:", ", ".join(pendientes))

print("clientes:", clientes.shape, "  boletas:", boletas.shape)
verificar("filas_clientes", len(clientes))
verificar("filas_boletas", len(boletas))
````

## Sección 1 · Corregir texto y categorías
*En Excel: `ESPACIOS`, `NOMPROPIO`, y `Ctrl+H` para los reemplazos dirigidos (guía de Excel, §4).*

`comuna` en `clientes` tiene **69 formas de escribir** unas pocas comunas reales (`CHILLÁN`,
`Chillan`, `chillán `…). `medio_pago` en `boletas` tiene **13 formas** de escribir **4** medios de
pago: *Débito, Efectivo, Crédito, Tarjeta Vecino*.

**Debe quedar:** `comuna` con **21 valores distintos** y `medio_pago` con
**4**. Modifique las columnas dentro de los mismos DataFrames
`clientes` y `boletas`.

### 🔎 Cómo se resuelve

**Qué se pide.** Que `comuna` quede con **21** valores distintos (hoy tiene 69) y `medio_pago` con **4** (hoy tiene 13).

**En Excel fue:** `ESPACIOS`, `NOMPROPIO` y `Ctrl+H` (guía de Excel, §4).

**La idea.** Primero se arregla lo **general** (espacios y mayúsculas, para todas las filas a la vez) y después lo **particular** (las tildes y las abreviaturas, con un diccionario *así viene → así debe quedar*). Es exactamente el orden que siguió en Excel.

**Línea por línea** (lo nuevo de esta sección):

| Código | Qué hace |
|---|---|
| `clientes.comuna` | la columna `comuna` del DataFrame `clientes`. También se escribe `clientes["comuna"]`; para **asignar** una columna se usa siempre la forma con corchetes |
| `.nunique()` | cuenta cuántos valores **distintos** hay (no cuenta los vacíos) |
| `.str.strip()` | `.str` da acceso a las funciones de texto; `strip()` quita los espacios del inicio y del final (`"chillán "` → `"chillán"`) |
| `.str.title()` | Mayúscula Inicial En Cada Palabra, como `NOMPROPIO` |
| `clientes["comuna"] = ...` | **reemplaza** la columna por la versión corregida. Si no asigna, pandas calcula el resultado, lo muestra y lo olvida |
| `{"Chillan": "Chillán", ...}` | un **diccionario**: pares `clave: valor`. Aquí, *así viene: así debe quedar* |
| `.replace(tildes)` | cambia solo los valores que aparecen como clave; el resto queda igual |
| `.map(canon)` | traduce **cada** valor con el diccionario; lo que no encuentra lo deja vacío (`NaN`). Por eso el control del paso 5 |
| `sorted(... .dropna().unique())` | la lista de comunas que quedaron, sin vacíos y en orden alfabético: para *mirarla* y confirmar que son 21 comunas reales |

````python
# SECCIÓN 1 · Corregir texto: comuna (clientes) y medio_pago (boletas)

# 1) Mirar antes de corregir: cuántas formas distintas hay hoy
print("comuna antes:", clientes.comuna.nunique(), "valores distintos")

# 2) Quitar espacios sobrantes y dejar Mayúscula Inicial (ESPACIOS + NOMPROPIO)
clientes["comuna"] = clientes.comuna.str.strip().str.title()
print("tras strip + title:", clientes.comuna.nunique())

# 3) Las que siguen distintas solo por la tilde: un diccionario (el Ctrl+H de Excel)
tildes = {"Chillan": "Chillán", "Chillan Viejo": "Chillán Viejo", "Quillon": "Quillón"}
clientes["comuna"] = clientes.comuna.replace(tildes)
print("comuna después:", clientes.comuna.nunique())
print(sorted(clientes.comuna.dropna().unique()))

# 4) medio_pago: 13 formas de escribir 4 medios. Diccionario {como viene: como debe quedar}
print(boletas.medio_pago.value_counts())
canon = {
    "Débito": "Débito", "debito": "Débito", "DEBITO": "Débito", "TD": "Débito",
    "Efectivo": "Efectivo", "EFECTIVO": "Efectivo", "ef": "Efectivo",
    "Crédito": "Crédito", "CREDITO": "Crédito", "TC": "Crédito",
    "Tarjeta Vecino": "Tarjeta Vecino", "T. Vecino": "Tarjeta Vecino",
    "TARJETA VECINO": "Tarjeta Vecino",
}
boletas["medio_pago"] = boletas.medio_pago.map(canon)

# 5) Control: lo que no esté en el diccionario, .map() lo deja vacío (NaN). Debe dar 0
print("vacíos tras el mapeo:", boletas.medio_pago.isna().sum())
print(boletas.medio_pago.value_counts())
````

## Sección 2 · Rellenar valores faltantes
*En Excel: `Ir a Especial → Celdas en blanco`, y luego `MODA.UNO` o escribir `"Sin dato"` (§5).*

Dos columnas de `clientes` tienen vacíos, y en Excel les dio **tratamientos distintos**. Repítalos:

- `comuna` → marcar explícitamente con `"Sin dato"` (no se puede adivinar dónde vive alguien).
- `canal_alta` → rellenar con el valor más frecuente (la **moda**).

**Antes de rellenar, cuente** cuántos vacíos hay en cada una. **Después**, no debe quedar ninguno.

### 🔎 Cómo se resuelve

**Qué se pide.** Contar los vacíos **antes** de rellenar (87 en `comuna`, 49 en `canal_alta`) y dejar cero vacíos después: `comuna` con `"Sin dato"`, `canal_alta` con la moda.

**En Excel fue:** `Ir a Especial → Celdas en blanco`, `MODA.UNO` o escribir `"Sin dato"` (§5).

**La idea.** Dos columnas, dos tratamientos: **no se inventa** dónde vive una persona (se declara el vacío), pero sí se puede suponer el canal más común para un dato poco crítico. Y se cuenta **antes**, porque después de rellenar ya no queda nada que contar.

**Línea por línea** (lo nuevo de esta sección):

| Código | Qué hace |
|---|---|
| `.isna()` | marca `True` en cada celda vacía y `False` en las demás |
| `.isna().sum()` | suma esos `True` (cada uno vale 1): el número de vacíos |
| `.fillna("Sin dato")` | rellena **solo** las celdas vacías con ese texto |
| `.mode()[0]` | `.mode()` devuelve una **lista** de modas (puede haber empate); `[0]` toma la primera |
| `clientes[["comuna", "canal_alta"]]` | doble corchete: una tabla con **varias** columnas (con un corchete sería una sola) |

````python
# SECCIÓN 2 · Faltantes en clientes: comuna → "Sin dato", canal_alta → moda

# 1) Contar ANTES de rellenar (después los vacíos ya no existen)
comunas_vacias = clientes.comuna.isna().sum()
canal_alta_vacios = clientes.canal_alta.isna().sum()
print("vacíos antes →  comuna:", comunas_vacias, "  canal_alta:", canal_alta_vacios)

# 2) comuna: no se puede adivinar dónde vive alguien → se declara el vacío
clientes["comuna"] = clientes.comuna.fillna("Sin dato")

# 3) canal_alta: se rellena con el valor más frecuente (MODA.UNO en Excel)
moda = clientes.canal_alta.mode()[0]
print("moda de canal_alta:", moda)
clientes["canal_alta"] = clientes.canal_alta.fillna(moda)

# 4) Control: no debe quedar ningún vacío en esas dos columnas
print(clientes[["comuna", "canal_alta"]].isna().sum())
````

## Sección 3 · Quitar duplicados
*En Excel: `Datos → Quitar duplicados`, eligiendo qué columnas definen el duplicado (§6).*

**Clientes:** la empresa dice tener 1.200 socios y el archivo trae 1.245
filas. Hay personas cargadas dos veces con el RUT escrito distinto (`90.147.265-3` y
`90147265-3`). Cree una columna `rut_norm` **sin puntos, sin guion y en mayúsculas**, y cuente
duplicados por `rut` tal cual y por `rut_norm`. Luego elimine los duplicados por `rut_norm`
**conservando la primera aparición**.

**Boletas:** hay boletas repetidas (`B0001593` aparece dos veces). Elimine los duplicados por
`id_boleta`, conservando la primera.

**Debe quedar:** 1.200 clientes y 57.467 boletas.

### 🔎 Cómo se resuelve

**Qué se pide.** Clientes por RUT normalizado (1.245 → **1.200**) y boletas por `id_boleta` (57.527 → **57.467**), conservando la primera aparición.

**En Excel fue:** `Datos → Quitar duplicados`, eligiendo las columnas (§6).

**La idea.** Un duplicado se define por **lo que identifica a la entidad**: a la persona la identifica el RUT, pero el RUT viene escrito de dos formas (`90.147.265-3` y `90147265-3`). Primero se lleva a una forma única y después se compara.

**Línea por línea** (lo nuevo de esta sección):

| Código | Qué hace |
|---|---|
| `.str.replace(".", "", regex=False)` | cambia cada `.` por nada. `regex=False` le dice que el punto es un punto (en las *expresiones regulares* el punto significa "cualquier carácter") |
| `( ... )` | los paréntesis permiten partir una instrucción larga en varias líneas |
| `.str.upper()` | a mayúsculas: el dígito verificador `k` y `K` quedan iguales |
| `.duplicated()` | `True` en cada fila cuyo valor **ya apareció antes**; la primera aparición queda en `False` |
| `len(clientes)` | cuántas filas tiene la tabla |
| `drop_duplicates(subset="rut_norm", keep="first")` | borra las filas repetidas mirando **solo** `rut_norm` y conserva la primera. `keep` vuelve en la G1 |
| `clientes = clientes.drop_duplicates(...)` | `drop_duplicates` entrega una tabla **nueva**; si no la guarda con `clientes = ...`, no pasó nada |
| `f"clientes: {antes} → {len(clientes)}"` | un *f-string*: lo que va entre `{ }` se reemplaza por su valor |

````python
# SECCIÓN 3 · Duplicados: clientes por rut_norm, boletas por id_boleta

# 1) RUT normalizado: sin puntos, sin guion, en mayúsculas (90.147.265-3 → 901472653)
clientes["rut_norm"] = (clientes.rut.str.replace(".", "", regex=False)
                                    .str.replace("-", "", regex=False)
                                    .str.upper())

# 2) Contar repetidos de las dos formas, para ver la diferencia
rut_repetidos_exactos = clientes.rut.duplicated().sum()
rut_repetidos_normalizados = clientes.rut_norm.duplicated().sum()
print("RUT repetidos → tal cual:", rut_repetidos_exactos,
      "  normalizados:", rut_repetidos_normalizados)

# 3) Eliminar duplicados por rut_norm, conservando la PRIMERA aparición
antes = len(clientes)
clientes = clientes.drop_duplicates(subset="rut_norm", keep="first")
print(f"clientes: {antes} → {len(clientes)}")

# 4) Boletas: el duplicado se define por id_boleta
boletas_repetidas_por_id = boletas.id_boleta.duplicated().sum()
antes = len(boletas)
boletas = boletas.drop_duplicates(subset="id_boleta", keep="first")
print(f"boletas: {antes} → {len(boletas)}  (se fueron {boletas_repetidas_por_id})")
````

## Sección 5 · Cruzar dos tablas
*En Excel: `BUSCARV(...; FALSO)` envuelto en `SIERROR(...; "NO EXISTE")` (§8).*

Traiga a cada boleta el `nombre` y la `comuna` de su cliente. Guarde el resultado en un
DataFrame `cruce`. Luego cuente dos cosas distintas:

- **boletas sin tarjeta**: `id_cliente` vacío en la boleta (compra anónima, no es un error);
- **boletas huérfanas**: tienen un `id_cliente`, pero **ese cliente no existe** en `clientes`
  (el `"NO EXISTE"` de Excel). Guárdelas en `huerfanas`.

### 🔎 Cómo se resuelve

**Qué se pide.** Traer a cada boleta el `nombre` y la `comuna` de su cliente, en `cruce`, y contar **22.812** boletas sin tarjeta y **90** huérfanas.

**En Excel fue:** `BUSCARV(...; FALSO)` dentro de `SIERROR(...; "NO EXISTE")` (§8).

**La idea.** Un cruce **izquierdo** conserva todas las boletas, encuentren o no a su cliente. Así se pueden *ver* las que fallan. Hay dos tipos de "sin cliente" y no son lo mismo: la compra **anónima** (no trae tarjeta: normal) y la **huérfana** (trae un `id_cliente` que no existe: error de integridad).

**Línea por línea** (lo nuevo de esta sección):

| Código | Qué hace |
|---|---|
| `boletas.merge(tabla, on="id_cliente", how="left")` | une las dos tablas por la columna común. `left`: manda la tabla de la izquierda (`boletas`); lo que no encuentra en `clientes` queda vacío |
| `clientes[["id_cliente", "nombre", "comuna"]]` | solo las columnas que se quieren traer (más la llave del cruce) |
| `.notna()` | lo contrario de `.isna()`: `True` donde **sí** hay dato |
| `cond1 & cond2` | `&` significa **y**: las dos condiciones a la vez |
| `[:4]` | los cuatro primeros elementos de una lista |

````python
# SECCIÓN 5 · Cruzar boletas con clientes y encontrar huérfanas

# 1) Cruce izquierdo: TODAS las boletas se quedan, encuentren o no a su cliente
cruce = boletas.merge(clientes[["id_cliente", "nombre", "comuna"]],
                      on="id_cliente", how="left")

# 2) Sin tarjeta: la boleta no trae id_cliente (compra anónima, no es un error)
sin_tarjeta = cruce.id_cliente.isna().sum()

# 3) Huérfana: trae id_cliente, pero el cruce no encontró su nombre ("NO EXISTE")
huerfanas = cruce[cruce.id_cliente.notna() & cruce.nombre.isna()]
print("boletas sin tarjeta:", sin_tarjeta)
print("boletas huérfanas:", len(huerfanas))
print("ids de ejemplo:", sorted(huerfanas.id_cliente.unique())[:4])
huerfanas[["id_boleta", "fecha_hora", "id_cliente", "total"]].head()
````

## Sección 6 · Derivar atributos nuevos
*En Excel: `=TEXTO(A2;"dddd")` y `=SIFECHA(...)` (§9).*

**Generar datos es derivar, no inventar.** Cree dos columnas nuevas:

- en `cruce`, `dia_semana` a partir de `fecha_hora`, **en español y en minúsculas**
  (`"lunes"`, …, `"domingo"`);
- en `clientes`, `edad` a partir de `fecha_nacimiento`, calculada como **2025 − año de nacimiento**.

Luego responda: ¿qué día de la semana tiene más boletas, y cuántas?

### 🔎 Cómo se resuelve

**Qué se pide.** Crear `dia_semana` (en español, minúsculas) en `cruce` y `edad` (2025 − año de nacimiento) en `clientes`. El día con más boletas es el **sábado**, con **12.658**.

**En Excel fue:** `=TEXTO(A2;"dddd")` y `=SIFECHA(...)` (§9).

**La idea.** **Generar datos es derivar, no inventar**: el día de la semana y la edad ya están *dentro* de las fechas; solo hay que sacarlos. Para eso, primero el texto `"2025-02-11 09:42:00"` tiene que convertirse en una fecha de verdad.

**Línea por línea** (lo nuevo de esta sección):

| Código | Qué hace |
|---|---|
| `pd.to_datetime(...)` | convierte texto en fecha: desde ahí se le puede pedir el año, el mes, el día de la semana… |
| `.dt.dayofweek` | `.dt` da acceso a las partes de una fecha; `dayofweek` es el día como número: 0 = lunes … 6 = domingo |
| `dict(enumerate(dias))` | numera la lista y la vuelve diccionario `{0: "lunes", …}`; con `.map()` cada número se traduce a su nombre |
| `.value_counts()` | cuántas veces aparece cada valor, de mayor a menor |
| `.idxmax() / .max()` | **cuál** es el mayor / **cuánto** vale |
| `errors="coerce"` | si una fecha no se puede leer, en vez de detenerse con error la deja vacía (`NaT`, *Not a Time*) |
| `.describe().round(0)` | resumen estadístico (cantidad, media, mínimo, cuartiles, máximo), sin decimales |

````python
# SECCIÓN 6 · Derivar dia_semana (cruce) y edad (clientes)

# 1) Día de la semana: texto → fecha → número 0-6 → nombre en español
dias = ["lunes", "martes", "miércoles", "jueves", "viernes", "sábado", "domingo"]
fechas = pd.to_datetime(cruce.fecha_hora)
# dict(enumerate(dias)) es {0: "lunes", 1: "martes", ..., 6: "domingo"}
cruce["dia_semana"] = fechas.dt.dayofweek.map(dict(enumerate(dias)))

# 2) Contar boletas por día y ver cuál es el mayor
por_dia = cruce.dia_semana.value_counts()
print(por_dia)
print("día con más boletas:", por_dia.idxmax(), "→", por_dia.max())

# 3) Edad = 2025 − año de nacimiento. Lo que no se pueda leer como fecha queda vacío (NaT)
nac = pd.to_datetime(clientes.fecha_nacimiento, errors="coerce")
clientes["edad"] = 2025 - nac.dt.year
print(clientes.edad.describe().round(0))
````

````python
# PREGUNTA G1 · keep="first" vs keep="last": ¿cuántas huérfanas quedan con cada uno?

# 1) Normalizar el RUT del crudo, igual que en la Sección 3
clientes_crudo["rut_norm"] = (clientes_crudo.rut.str.replace(".", "", regex=False)
                                                .str.replace("-", "", regex=False)
                                                .str.upper())

# 2) Dos versiones deduplicadas: conservando la primera copia o la última
clientes_first = clientes_crudo.drop_duplicates(subset="rut_norm", keep="first")
clientes_last = clientes_crudo.drop_duplicates(subset="rut_norm", keep="last")

# 3) ¿Qué ids sobreviven? Mire con qué letras empiezan (C0... o C9...)
print("first:", clientes_first.id_cliente.str[:2].value_counts().to_dict())
print("last: ", clientes_last.id_cliente.str[:2].value_counts().to_dict())

# 4) Huérfana = boleta con tarjeta cuyo cliente NO está entre los conservados
con_tarjeta = boletas.id_cliente.notna()
huerfanas_first = (con_tarjeta & ~boletas.id_cliente.isin(clientes_first.id_cliente)).sum()
huerfanas_last = (con_tarjeta & ~boletas.id_cliente.isin(clientes_last.id_cliente)).sum()
print("huérfanas con first:", huerfanas_first, "   con last:", huerfanas_last)
````

**La idea.** Un algoritmo no lee dos tablas: lee **una fila por caso**, con sus atributos y su clase. Todo sale de lo que usted ya limpió: la edad de la Sección 6 (con los blancos de la G3), la comuna de la Sección 1, el canal de la Sección 2 y la última compra de `cruce`. **Fugado** = sin compras desde el 01/05/2025.

**Línea por línea** (lo nuevo de esta sección):

| Código | Qué hace |

| `.dropna(subset=["id_cliente"])` | descarta las boletas sin tarjeta: no tienen a quién asignarse |

````python
# SECCIÓN 9 · Armar la tabla de modelado (una fila por cliente)

# 1) Las seis comunas que tienen local (un conjunto, entre llaves)
LOCAL = {"Chillán", "Chillán Viejo", "San Carlos", "Bulnes", "Coihueco", "Quillón"}

# 2) Fecha de la última compra de cada cliente, desde las boletas de venta
con_cliente = cruce.dropna(subset=["id_cliente"])
ultima = pd.to_datetime(con_cliente.groupby("id_cliente").fecha_hora.max())

# 3) De edad a tramo. Las edades en blanco (G3) van a "Sin dato"
def tramo(edad):
    if pd.isna(edad):
        return "Sin dato"
    if edad <= 29:
        return "18-29"
    if edad <= 44:
        return "30-44"
    if edad <= 59:
        return "45-59"
    return "60+"

# 4) Traductor de True/False a "Sí"/"No"
si_no = {True: "Sí", False: "No"}

# 5) Fugado = su última compra es anterior al 01/05/2025
fugado = clientes.id_cliente.map(ultima) < pd.Timestamp("2025-05-01")

# 6) La tabla de modelado: una fila por cliente, cuatro atributos y la clase
modelo = pd.DataFrame({
    "tramo":       clientes.edad.map(tramo),
    "promos":      clientes.acepta_promociones.map(si_no),
    "local_cerca": clientes.comuna.isin(LOCAL).map(si_no),
    "canal_alta":  clientes.canal_alta,
    "fuga":        fugado.map(si_no),
})
print(modelo.shape)
print(modelo.fuga.value_counts())
modelo.head()
````

## Sección 10 · ZeroR: la regla sin atributos

**ZeroR** es el modelo más tonto posible: mira solo la clase, elige la mayoritaria y la responde
siempre. Es la **vara mínima**: cualquier modelo que no le gane, no sirve.

Calcule cuál es la clase mayoritaria, cuántos casos acierta ZeroR y qué porcentaje es.

### 🔎 Cómo se resuelve

**Qué se pide.** La clase mayoritaria y cuántos casos acierta ZeroR (**909** de 1.200 = 75,8 %).

**En Excel fue:** el primer paso de la actividad a mano: *¿qué clase es la mayoría?*.

**La idea.** ZeroR no mira ningún atributo: responde siempre la clase más frecuente. Es la **vara mínima**: un modelo que no le gana a ZeroR no aporta nada.

**Línea por línea** (lo nuevo de esta sección):

| Código | Qué hace |
|---|---|
| `conteo.idxmax() / conteo.max()` | la clase más frecuente / cuántas veces aparece |
| `:.1%` | dentro de un f-string, formatea como porcentaje con un decimal |

````python
# SECCIÓN 10 · ZeroR sobre los 1.200 clientes

# ZeroR: mirar solo la clase y responder siempre la mayoritaria
conteo = modelo.fuga.value_counts()
print(conteo)
regla_zeror = conteo.idxmax()       # la clase que más se repite
zeror_aciertos = conteo.max()       # cuántas veces acierta si responde siempre esa
print(f'ZeroR dice siempre "{regla_zeror}" y acierta {zeror_aciertos} de {len(modelo)}',
      f"= {zeror_aciertos / len(modelo):.1%}")
````

## Sección 11 · OneR con un atributo

**OneR** (Holte, 1993) mira **un solo atributo**: para cada uno de sus valores elige la clase
mayoritaria dentro de ese valor, y esa lista de "si valor → clase" es el modelo. Igual que la
tabla que llenó a mano el 21/09.

Empiece por `tramo`: construya la tabla de frecuencias (valores × clase), léala, y cuente cuántos
casos acierta la regla.

### 🔎 Cómo se resuelve

**Qué se pide.** La tabla `tramo` × `fuga`, la regla de cada tramo y los aciertos (**909**).

**En Excel fue:** la tabla de conteo que llenó a mano el 21/09.

**La idea.** `pd.crosstab` arma **la misma tabla que usted llenó a mano**: una fila por valor del atributo, una columna por clase. En cada fila, la regla elige la clase que más se repite, y acierta exactamente ese conteo.

**Línea por línea** (lo nuevo de esta sección):

| Código | Qué hace |
|---|---|
| `pd.crosstab(filas, columnas)` | tabla de contingencia: cuenta cuántos casos hay en cada combinación |
| `axis=1` | *por fila* (`axis=0` sería por columna) |
| `for valor, clase in ….items():` | recorre la regla fila a fila; el `print` de adentro va con sangría |
| `{valor:9s}` | escribe el texto ocupando 9 espacios, para que la regla quede alineada |
| `"\n"` | salto de línea |

````python
# SECCIÓN 11 · OneR con el atributo tramo

# 1) Tabla de frecuencias: una fila por tramo, una columna por clase
tabla = pd.crosstab(modelo.tramo, modelo.fuga)
print(tabla)

# 2) La regla: en cada fila, la clase que más se repite
print("\nLa regla que propone OneR:")
for valor, clase in tabla.idxmax(axis=1).items():
    print(f"  si tramo = {valor:9s} → fuga = {clase}")

# 3) Aciertos: en cada fila se acierta el conteo mayor; se suman todas las filas
oner_tramo_aciertos = tabla.max(axis=1).sum()
print(f"\naciertos: {oner_tramo_aciertos} de {len(modelo)}",
      f"= {oner_tramo_aciertos / len(modelo):.1%}")
````

## Sección 12 · OneR con los cuatro atributos

OneR se llama así porque **prueba todos los atributos y se queda con el mejor**. Repita la Sección
11 para los cuatro (`tramo`, `promos`, `local_cerca`, `canal_alta`) y arme la comparativa:
atributo, aciertos, y si le gana o no a ZeroR.

### 🔎 Cómo se resuelve

**Qué se pide.** Repetir la Sección 11 con los cuatro atributos y ver cuál le gana a ZeroR (mejor: **909**; atributos que ganan: **0**).

**En Excel fue:** en la actividad a mano, se repitió la tabla por atributo.

**La idea.** OneR se llama así porque **prueba todos los atributos y se queda con el mejor**. Un `for` repite la Sección 11 para cada uno. **Mire bien el resultado antes de pensar que se equivocó.**

**Línea por línea** (lo nuevo de esta sección):

| Código | Qué hace |
|---|---|
| `aciertos = {}` | un diccionario vacío, que el `for` va llenando: `{"tramo": …, "promos": …}` |
| `modelo[col]` | la columna cuyo nombre está en la variable `col` (aquí se usan corchetes porque el nombre cambia en cada vuelta) |
| `comparativa.aciertos > zeror_aciertos` | `True` si ese atributo le gana a ZeroR |
| `["¿le gana a ZeroR?"].sum()` | cuenta los `True`: cuántos atributos ganan |

````python
# SECCIÓN 12 · Los cuatro atributos, y cuál gana

# 1) Repetir la Sección 11 con cada atributo y guardar sus aciertos
aciertos = {}
for col in ["tramo", "promos", "local_cerca", "canal_alta"]:
    t = pd.crosstab(modelo[col], modelo.fuga)
    aciertos[col] = t.max(axis=1).sum()

# 2) Comparativa: aciertos, porcentaje y si le gana a ZeroR
comparativa = pd.DataFrame({"aciertos": aciertos})
comparativa["% acierto"] = (comparativa.aciertos / len(modelo) * 100).round(1)
comparativa["¿le gana a ZeroR?"] = comparativa.aciertos > zeror_aciertos
print("ZeroR:", zeror_aciertos, "aciertos\n")
print(comparativa)

# 3) Los dos números que se verifican
oner_mejor_aciertos = comparativa.aciertos.max()
oner_atributos_que_ganan = comparativa["¿le gana a ZeroR?"].sum()
print("\nMejor OneR:", oner_mejor_aciertos,
      "· atributos que le ganan a ZeroR:", oner_atributos_que_ganan)
````

## Sección 13 · La señal que OneR no vio

Ningún atributo le ganó a ZeroR. ¿Significa que la edad no dice nada sobre la fuga? Mírelo de
otra forma: en vez de la **clase mayoritaria** de cada valor, calcule el **porcentaje de fugados**
dentro de cada valor.

**Debe quedar:** de los **220** clientes de `60+` se fugaron **8**;
de los **272** de `18-29`, **106**.

### 🔎 Cómo se resuelve

**Qué se pide.** El **porcentaje** de fugados dentro de cada tramo: `60+` **8 de 220**; `18-29` **106 de 272**.

**En Excel fue:** calcular el porcentaje dentro de cada fila de la tabla.

**La idea.** OneR pregunta *¿cuál es la clase mayoritaria en cada valor?* Aquí se pregunta otra cosa: *¿qué porcentaje se fuga en cada valor?* Son preguntas distintas, y la segunda sí ve diferencias.

**Línea por línea** (lo nuevo de esta sección):

| Código | Qué hace |
|---|---|
| `normalize="index"` | en vez de conteos, proporciones **por fila** |
| `.mul(100).round(1)` | multiplica por 100 y deja un decimal: porcentajes |
| `(modelo.tramo == "60+")` | `==` **compara** (un solo `=` **asigna**) |

````python
# SECCIÓN 13 · Tasa de fuga por valor, no clase mayoritaria

# 1) Porcentaje de fugados DENTRO de cada tramo (normalize="index": cada fila suma 100)
tasas = pd.crosstab(modelo.tramo, modelo.fuga, normalize="index").mul(100).round(1)
print(tasas)

# 2) Los cuatro conteos que pide la verificación
clientes_60mas = (modelo.tramo == "60+").sum()
fugados_60mas = ((modelo.tramo == "60+") & (modelo.fuga == "Sí")).sum()
clientes_18_29 = (modelo.tramo == "18-29").sum()
fugados_18_29 = ((modelo.tramo == "18-29") & (modelo.fuga == "Sí")).sum()
print(f"\n60+  : {fugados_60mas} de {clientes_60mas} se fugaron")
print(f"18-29: {fugados_18_29} de {clientes_18_29} se fugaron")

# 3) Lo mismo con promos: ¿se fuga más quien acepta promociones o quien no?
pd.crosstab(modelo.promos, modelo.fuga, normalize="index").mul(100).round(1)
````

### 🔎 Comprobación opcional de G4

Si el empate se debe al desbalance, con una muestra **balanceada** (todos los fugados y la misma cantidad de no fugados, elegidos al azar) OneR debería ganarle a ZeroR. `.sample(n, random_state=2026)` elige `n` filas al azar, siempre las mismas, y `pd.concat` apila las dos tablas.

````python
# G4 · comprobación opcional: todos los fugados + la misma cantidad de no fugados al azar
fugados = modelo[modelo.fuga == "Sí"]
no_fugados = modelo[modelo.fuga == "No"].sample(len(fugados), random_state=2026)
balanceada = pd.concat([fugados, no_fugados])
print("filas:", len(balanceada), "· ZeroR acierta:", balanceada.fuga.value_counts().max())

t = pd.crosstab(balanceada.tramo, balanceada.fuga)
print(t)
print("OneR con tramo acierta:", t.max(axis=1).sum(), "de", len(balanceada))
````
